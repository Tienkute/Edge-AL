### 1 & 2. Đọc file `intent_train.csv` và in 5 dòng đầu tiên

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

# Thiết lập hiển thị đồ thị inline
%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11

# Xác định đường dẫn file
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
INTENT_TRAIN_PATH = PROJECT_ROOT / 'data' / 'raw' / 'intent_train.csv'
INTENT_TEST_PATH = PROJECT_ROOT / 'data' / 'raw' / 'intent_test.csv'

# Đọc file train CSV bằng Pandas
df_train = pd.read_csv(INTENT_TRAIN_PATH)
print("5 dòng đầu tiên của tập train:")
df_train.head()
```

### 3. Hiển thị kích thước dataset

```python
print(f"Kích thước tập huấn luyện (shape): {df_train.shape}")
print(f"Số lượng câu hỏi train:           {len(df_train)}")
print(f"Danh sách các cột:                {list(df_train.columns)}")
```

### 4. Liệt kê các Intent khác nhau trong dataset

```python
unique_intents = sorted(df_train['intent'].unique())
print(f"Tổng số nhãn intent phân biệt: {len(unique_intents)}\n")
for i, intent in enumerate(unique_intents, start=1):
    print(f"  {i}. {intent}")
```

### 5. Đếm số câu của từng Intent

```python
intent_counts = df_train['intent'].value_counts()
df_intent_summary = pd.DataFrame({
    'Số câu': intent_counts,
    'Tỷ lệ (%)': (intent_counts / len(df_train) * 100).round(2)
})
print("Thống kê số lượng câu hỏi theo từng intent:")
print(df_intent_summary)
```

### 6. Vẽ biểu đồ Bar Chart phân bố Intent

```python
plt.figure(figsize=(10, 5))
bars = plt.bar(intent_counts.index, intent_counts.values, color='#3c8dbc', edgecolor='black', alpha=0.85)

# Hiển thị nhãn số lượng trên từng cột
for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2.0, h + 0.4, f"{int(h)}",
             ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title("Phân bố số lượng câu hỏi theo từng Intent trong tập Train", fontsize=13, fontweight='bold')
plt.xlabel("Intent", fontsize=11)
plt.ylabel("Số lượng câu hỏi", fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 7. Tính độ dài từng câu hỏi theo số từ (Word Count)

```python
# Đếm số từ bằng cách tách chuỗi theo khoảng trắng
df_train['word_count'] = df_train['question'].apply(lambda q: len(str(q).split()))

print("Thống kê mô tả về độ dài câu hỏi (số từ):")
print(df_train['word_count'].describe())

plt.figure(figsize=(9, 4.5))
plt.hist(df_train['word_count'], bins=12, color='#2ca02c', edgecolor='black', alpha=0.8)
plt.title("Phân bố độ dài câu hỏi theo số từ (Train set)", fontsize=13, fontweight='bold')
plt.xlabel("Số từ trong một câu", fontsize=11)
plt.ylabel("Số lượng câu", fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 8. In một vài câu ví dụ thực tế của mỗi Intent

```python
for intent in unique_intents:
    print(f"\n=== INTENT: {intent} ===")
    samples = df_train[df_train['intent'] == intent]['question'].head(3).tolist()
    for idx, q in enumerate(samples, start=1):
        print(f"  [{idx}] {q}")
```

### 9. Đọc `intent_test.csv` và kiểm tra số lượng câu hỏi

```python
df_test = pd.read_csv(INTENT_TEST_PATH)
print(f"Kích thước tập test (shape): {df_test.shape}")
print(f"Số lượng câu hỏi cần dự đoán: {len(df_test)}")
print(f"Các cột trong tập test:       {list(df_test.columns)}\n")
print("5 dòng đầu tiên của tập test:")
df_test.head()
```
